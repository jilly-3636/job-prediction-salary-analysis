```python
import pandas as pd
from scipy import stats



```

```python
features = ["experience_years", "skills_count", "certifications"]

sample = job_df.sample(5000, random_state=42)

for col in features + ["salary"]:
    stat, p = stats.shapiro(sample[col])
    print(col, "| p-value:", p, "->", "NOT normal" if p < 0.05 else "normal")
```

```python
for col in features:
    rho, p_value = stats.spearmanr(job_df[col], job_df["salary"])

    print("=" * 50)
    print(f"Feature: {col} vs salary")
    print(f"H0: There is no relationship between {col} and salary")
    print(f"H1: There is a relationship between {col} and salary")
    print(f"Correlation (rho): {rho:.3f}")
    print(f"P-value: {p_value:.4f}")

    if p_value < 0.05:
        print("Decision: p < 0.05, so REJECT H0. There is a significant relationship.")
    else:
        print("Decision: p >= 0.05, so FAIL TO REJECT H0. No significant relationship.")
```

```python
import scikit_posthocs as sp
from scipy import stats

def test_groups(col):
    groups = [g["salary"].values for _, g in job_df.groupby(col)]
    n, k = len(job_df), len(groups)

    print("=" * 60)
    print(f"Salary vs {col}")
    print(f"H0: Salary is the same across all {col} groups")
    print(f"H1: Salary differs in at least one {col} group")

    # Assumption 1: sample sizes
    print("\nSmallest group size:", min(len(g) for g in groups))

    # Assumption 2: normality inside each group (sample of 5000 for Shapiro)
    normal_flags = []
    for name, g in job_df.groupby(col):
        s = g["salary"].sample(min(len(g), 5000), random_state=42)
        normal_flags.append(stats.shapiro(s).pvalue >= 0.05)
    print("All groups normal?", all(normal_flags))

    # Assumption 3: equal spread (Levene)
    lev_p = stats.levene(*groups, center="median").pvalue
    print(f"Equal variances? Levene p-value = {lev_p:.4f} ->",
          "yes" if lev_p >= 0.05 else "no")

    # Main test: Kruskal-Wallis
    H, p_value = stats.kruskal(*groups)
    eps2 = (H - k + 1) / (n - k)
    print(f"\nKruskal-Wallis H = {H:.1f}")
    print(f"P-value: {p_value:.4f}")
    print(f"Effect size (eps2): {eps2:.4f}")

    print("\nMedian salary by group:")
    print(job_df.groupby(col)["salary"].median().sort_values())

    if p_value < 0.05:
        print("\nDecision: p < 0.05, so REJECT H0. Salary differs across groups.")
        print("Follow-up (Dunn's test, Bonferroni adjusted). "
              "p < 0.05 means those two groups differ:")
        dunn = sp.posthoc_dunn(job_df, val_col="salary", group_col=col,
                               p_adjust="bonferroni")
        print(dunn.round(4).to_string())
    else:
        print("\nDecision: p >= 0.05, so FAIL TO REJECT H0. No significant difference.")
```

```python
test_groups("education_level")   # B2.4
test_groups("company_size")      # B2.5
test_groups("remote_work")       # B2.6
test_groups("job_title")         # B2.7
test_groups("industry")          # B2.8
test_groups("location")          # B2.9
```

```python
import statsmodels.formula.api as smf

formula = ("salary ~ experience_years + skills_count + certifications"
           " + C(job_title) + C(education_level) + C(industry)"
           " + C(company_size) + C(location) + C(remote_work)")

model = smf.ols(formula, data=job_df).fit()

print("H0: None of the nine features help explain salary")
print("H1: At least one feature helps explain salary")
print(f"R-squared: {model.rsquared:.4f}")
print(f"F-statistic: {model.fvalue:.1f}")
print(f"P-value: {model.f_pvalue:.4f}")

if model.f_pvalue < 0.05:
    print("Decision: p < 0.05, so REJECT H0. The model is significant.")
else:
    print("Decision: p >= 0.05, so FAIL TO REJECT H0.")
```

```python
from statsmodels.stats.anova import anova_lm

anova = anova_lm(model, typ=2)
anova["pEta2"] = anova["sum_sq"] / (anova["sum_sq"] + anova.loc["Residual", "sum_sq"])
anova = anova.drop("Residual").rename(columns={"PR(>F)": "p_value"})

for feature, row in anova.iterrows():
    print("=" * 50)
    print(f"Feature: {feature}")
    print(f"H0: {feature} has no effect on salary (after accounting for the other features)")
    print(f"H1: {feature} has an effect on salary")
    print(f"F = {row['F']:.1f} | P-value: {row['p_value']:.4f} | pEta2: {row['pEta2']:.3f}")
    if row["p_value"] < 0.05:
        print("Decision: p < 0.05, so REJECT H0.")
    else:
        print("Decision: p >= 0.05, so FAIL TO REJECT H0.")
```

```python
from scipy import stats
from statsmodels.stats.diagnostic import het_breuschpagan

resid = model.resid

# 1. Are the leftover errors roughly normal?
print("Residual skew:", round(stats.skew(resid), 3))
print("Shapiro p-value:", stats.shapiro(resid.sample(5000, random_state=42)).pvalue)

# 2. Is the spread of errors constant? (Breusch-Pagan)
print("Breusch-Pagan p-value:", het_breuschpagan(resid, model.model.exog)[1])

# 3. Are features too similar to each other? (VIF)
import pandas as pd, statsmodels.api as sm
from statsmodels.stats.outliers_influence import variance_inflation_factor
X = sm.add_constant(pd.get_dummies(job_df.drop(columns="salary"), drop_first=True).astype(float))
vif = [variance_inflation_factor(X.values, i) for i in range(1, X.shape[1])]
print("Highest VIF:", round(max(vif), 2))

# 4. Is the experience relationship a straight line?
job_df.groupby("experience_years")["salary"].mean().plot(marker="o")
```
